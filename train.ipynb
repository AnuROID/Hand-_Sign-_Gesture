```python
import os
import cv2
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
data_dir = "C:/yora/handsignrecoprjkt/data/"
categories = os.listdir(data_dir)

data = []
labels = []

for category in categories:
    folder_path = os.path.join(data_dir, category)
    label = categories.index(category)  # Assign numeric labels

    for img_name in os.listdir(folder_path):
        img_path = os.path.join(folder_path, img_name)
        img = cv2.imread(img_path)
        img = cv2.resize(img, (64, 64))  # Resize to 64x64 for uniformity
        img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)  # Convert to grayscale
        img = img.flatten()  # Flatten image to 1D array
        data.append(img)
        labels.append(label)

data = np.array(data)
labels = np.array(labels)
```

```python
X_train, X_test, y_train, y_test = train_test_split(data, labels, test_size=0.2, random_state=42)
```

```python
model = SVC(kernel='linear')
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

```python
import joblib
joblib.dump(model, "hand_sign_model.pkl")
print("Model saved successfully!")
```
